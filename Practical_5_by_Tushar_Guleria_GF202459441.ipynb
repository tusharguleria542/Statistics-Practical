```python
df = pd.read_csv("/content/TeachingRatings.csv")
print(f"Total number of rating records: {len(df)}")  # quick sanity check – how many rows?
```

df['eval'] > 4.5 → checks each row, gives a column of True/False.

.mean() → since True=1 and False=0, averaging them = (count of True) ÷ (total rows) = the proportion.

print(...) → shows the result rounded to 4 decimals.

Result: 79/463 teachers scored above 4.5 → P ≈ 0.1706 (17.06%).

This is the empirical probability (straight from the data), vs. the earlier theoretical one (≈18.3%) which assumed a normal distribution. Close, but not exact — real data rarely fits the bell curve perfectly.

```python
# --------------------------------------------------------------
# 3️⃣  Q1 – probability that a teacher gets an eval score > 4.5
# --------------------------------------------------------------
mask_gt_45 = df['eval'] > 4.5          # True where the condition holds
prob_q1   = mask_gt_45.mean()           # mean of a Boolean array == proportion
print(f"\nQ1 – P(eval > 4.5) ≈ {prob_q1:.4f}")
```

(df['eval'] > 3.5) & (df['eval'] < 4.2) → two conditions checked per row, combined with & (AND) — True only if both hold.

.mean() → proportion of Trues = fraction of teachers in that range.

print(...) → shows result to 4 decimals.

Result: 164/463 teachers → P ≈ 0.3542 (35.42%)

(vs. ≈45.7% from the normal-distribution estimate — bigger gap this time, meaning the real data deviates more from a perfect bell curve in this range.)

```python
# --------------------------------------------------------------
# 4️⃣  Q2 – probability that eval is between 3.5 and 4.2 (strictly)
# --------------------------------------------------------------
mask_between = (df['eval'] > 3.5) & (df['eval'] < 4.2)
prob_q2     = mask_between.mean()
print(f"Q2 – P(3.5 < eval < 4.2) ≈ {prob_q2:.4f}")
```


print("\nHypotheses") → prints a blank line then the word "Hypotheses" as a header (\n = newline).

Second print → states H0: the regional players' mean score is not different from the historical mean of 12.

Third print → states Ha: the regional players' mean score is different from 12.

```python
# --------------------------------------------------------------
# 5️⃣  Q3 – Two‑tailed hypothesis test (basketball example)
# --------------------------------------------------------------
# 3.1  State the hypotheses
print("\nHypotheses")
print("H0 (null): The mean point of the regional players is NOT different from the historic mean (12).")
print("Ha (alt.): The mean point of the regional players IS different from the historic mean (12).")
```

Just setting up variables — storing the numbers from the problem so they can be used in the calculation later.

mu_hist = 12 → the historic mean points/game for pros (this is the value we're testing against, i.e. the H0 value).

sigma = 5.5 → the known population standard deviation (from the pros' historic data).

n = 36 → how many regional players were sampled.

x_bar = 10.7 → the average score actually observed in that sample of 36 regional players.

```python
# 3.2  Define given numbers
mu_hist = 12          # historic population mean for pros
sigma   = 5.5         # population standard deviation (known)
n       = 36          # sample size of regional players
x_bar   = 10.7        # observed sample mean
```

Output: z-score = -1.4182

sigma / np.sqrt(n) → this is the standard error — how much sample means of size 36 are expected to naturally vary from the true mean, just by chance.

x_bar - mu_hist → the actual gap between what we observed (10.7) and what we expected under H0 (12).

Dividing the two → converts that raw gap into a z-score: "how many standard errors away from 12 is our sample mean?"

Result is negative (-1.4182) because 10.7 is below 12 — the regional players scored lower than the historic pro mean, by about 1.4 standard errors.

print(f"...{z_score:.4f}") → displays it rounded to 4 decimal places.

```python
# 3.3  Compute the z‑score
#     z = (x̄ – μ) / (σ/√n)
z_score = (x_bar - mu_hist) / (sigma / np.sqrt(n))
print(f"\nz‑score = {z_score:.4f}")
```

Output: Two-tailed p-value = 0.1561

abs(z_score) → makes the z-score positive (-1.4182 → 1.4182), so we only need to work with one side of the curve.

stats.norm.sf(x) → "survival function" = 1 − CDF(x) = the area under the normal curve to the right of that z-value (i.e., the probability of getting a z-score that extreme or more, in one direction).

2 * → doubles it, because this is a two-tailed test — we need to count extreme values in both directions (much higher than 12 OR much lower than 12), not just one.

print(...) → shows the result rounded to 4 decimals.

The commented-out lines below are just an alternative way to write the same thing (norm.sf(...) directly instead of stats.norm.sf(...)) — depends on whether you imported norm directly (from scipy.stats import norm) or the whole stats module (from scipy import stats). Same result either way.

```python
# 3.4  Two‑tailed p‑value
p_value = 2 * stats.norm.sf(abs(z_score))   # sf(x) = 1 – CDF(x)
print(f"Two‑tailed p‑value = {p_value:.4f}")

# alternative method
#p_value = 2 * norm.sf(abs(z_score))   # sf(x) = 1 – CDF(x)
#print(f"Two‑tailed p‑value = {p_value:.4f}")
```

alpha = 0.05 → the significance threshold — the cutoff probability we use to decide "is this result surprising enough to matter?" (standard convention is 5%).

if p_value < alpha: → checks whether the p-value we calculated (0.1561) is smaller than 0.05.

Since 0.1561 is not less than 0.05, the else branch runs → prints "Fail to reject H0."

In plain terms: the p-value tells us there's about a 15.6% chance of seeing a sample mean this far from 12 just by random chance, even if the true regional mean really is 12. That's too high a chance to call it "surprising" (we need it below 5% to be confident it's a real difference), so we don't have enough evidence to say the regional players' average is statistically different from the pros' historic mean of 12.

```python
# 3.5  Decision at α = 0.05
alpha = 0.05
if p_value < alpha:
    print("Conclusion: Reject H0 – there is evidence the regional mean differs from the historic mean.")
else:
    print("Conclusion: Fail to reject H0 – no evidence that the regional mean differs from the historic mean.")
```
