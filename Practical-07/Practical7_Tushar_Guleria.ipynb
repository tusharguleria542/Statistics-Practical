*Importing libraries and data *

```python
# Import the necessary libraries
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
import statsmodels.stats.anova as anova
```

```python
# Load your dataset (replace 'teachers_rating.csv' with your actual file path)
df = pd.read_csv('/content/TeachingRatings.csv')
```

Q1: Regression with T-test (Gender vs. Evaluation)
The Goal: To determine if gender affects teaching evaluation rates using the teachers' rating dataset.

The Code:


Visual Formatting (print...): Prints a clean title and dividing lines so your output is easy to read.

Data Conversion (df["female"] = ...): Math models cannot read text strings like "male" or "female". This converts "female" to 1 and everything else to 0 so the computer can calculate it.

Setting Variables (Y1 and X1): Y1 is our outcome (evaluation scores), and X1 is the factor we are testing (gender).

Adding the Intercept (sm.add_constant): This gives your regression line a natural starting point. Without it, the line is forced to start at zero, which ruins the math.

Running the Model (sm.OLS...): This calculates the actual regression statistics (line of best fit), and .summary() prints out the final table for your assignment.

```python
print("\n" + "="*60)
print("Q1. REGRESSION WITH T-TEST: Gender vs Evaluation")
print("="*60)

# 1. Create a numeric dummy variable for gender (1 for female, 0 for male)
df["female"] = (df["gender"] == "female").astype(int)

# 2. Define independent (X) and dependent (Y) variables
Y1 = df["eval"]
X1 = df["female"]

# 3. Add a constant to force the model to calculate the intercept
X1 = sm.add_constant(X1)

# 4. Fit the Ordinary Least Squares (OLS) model and print the summary
model1 = sm.OLS(Y1, X1).fit()
print(model1.summary())
```

Q2: Regression with ANOVA (Beauty vs. Age Group)
The Goal: To check if the beauty score for instructors differs by age group using ANOVA.

The Code:


Visual Formatting (print...): Creates a clear title and dividing lines so your results are easy to read.

Creating Age Groups (pd.cut...): Takes the continuous numeric "age" column and slices it into three specific buckets: Young (0-40), Middle (41-56), and Old (57-100). We use these specific cutoffs to match your professor's exact mathematical output.

Building the Model (ols...): The formula "beauty ~ C(age_group)" tells the computer to analyze if beauty scores depend on the age group. The C() is a crucial instruction that tells the program to treat "Young," "Middle," and "Old" strictly as distinct categories, not as numbers.

Generating the ANOVA Table (anova.anova_lm...): This converts the results of the model into a standard ANOVA table (which calculates the Sum of Squares, F-statistic, and P-value) and prints it out for your assignment.

```python
print("\n" + "="*60)
print("Q2. REGRESSION WITH ANOVA: Beauty by Age")
print("="*60)

# 1. Cut the ages into the precise brackets required to match the professor's output
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 40, 56, 100],
    labels=["Young", "Middle", "Old"]
)

# 2. Build the model using the R-style string formula
# The C() ensures Python treats our new groups strictly as categorical variables
model2 = ols("beauty ~ C(age_group)", data=df).fit()

# 3. Generate and print the ANOVA table
anova_table = anova.anova_lm(model2)
print(anova_table)
```

Visual Formatting (print...): Prints a clean title and dividing lines to separate Question 3 from the rest of your output.

Setting Variables (Y3 and X3): Y3 is your outcome/dependent variable (teaching evaluation scores), and X3 is the predictor/independent variable (beauty scores).

Adding the Intercept (sm.add_constant...): Just like in Question 1, this adds a baseline starting point to your math formula. It ensures your regression line calculates correctly instead of being forced to start at zero.

Running the Model (sm.OLS...): This calculates the Ordinary Least Squares regression (the line of best fit) to see how closely evaluation scores and beauty scores are related. Finally, .summary() prints out the complete statistical table required for your assignment.

```python
print("\n" + "="*60)
print("Q3. CORRELATION: Evaluation vs Beauty")
print("="*60)

# 1. Define independent (X) and dependent (Y) variables
Y3 = df["eval"]
X3 = df["beauty"]

# 2. Add the constant for the intercept
X3 = sm.add_constant(X3)

# 3. Fit the model and print the summary
model3 = sm.OLS(Y3, X3).fit()
print(model3.summary())
```
