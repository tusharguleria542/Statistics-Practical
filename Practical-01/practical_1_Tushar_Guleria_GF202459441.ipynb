Importing pandas and numpy

```python
import pandas as pd
import numpy as np


```

STEP 1: Take raw data and converting to DataFrame using pandas function

df[...]: This tells Pandas to look inside your DataFrame (df) and select only the specified columns: gender, parental level of education, math score, and reading score.

.head(): This is a built-in Pandas function that returns the first 5 rows of the DataFrame by default, allowing you to quickly preview your data and verify that it loaded successfully.

print("\n"): This prints an empty blank line to the console (\n is the standard newline character), ensuring your upcoming outputs have neat spacing and are easy to read.


```python

# Load the CSV file you just uploaded to Colab
df = pd.read_csv('/content/StudentsPerformance.csv')

print(" STEP 1: Loading the dataframe ")
print(df[['gender', 'parental level of education', 'math score', 'reading score']].head())
print("\n")
```

 STEP 2: Using numpy find out: shape, size, dimension

 Extracting a 2x2 numerical matrix (Math and Reading scores of the first 2 students)

 I am making 2x2 matrix as :

 1.In linear algebra, you can only calculate the mathematical inverse of a square matrix—meaning the number of rows must perfectly equal the number of columns.

2.Preventing an Error Crash
If you extracted 2 columns but let Pandas grab the default 5 rows (creating a 5x2 rectangle), the np.linalg.inv() function would instantly fail. Numpy would throw a LinAlgError: Last 2 dimensions of the array must be square error, and your script would crash.

3 . It Makes Matrix Multiplication (Dot Product) Easy
To multiply two matrices together (the Dot Product in Step 3), the inner dimensions of the matrices must match. By ensuring both Matrix A and Matrix B are neatly packed 2x2 squares, you guarantee that all of the element-wise math (addition, subtraction, multiplication) and the algebraic dot product will execute perfectly without dimension mismatch errors.


```python

matrix_a = df[['math score', 'reading score']].head(2).to_numpy()

print(" STEP 2: using numpy to make a matrix ")
print(f"Matrix A:\n{matrix_a}")

```

Now using nnumpy to find out shape , size and dimension

```python
print(f"Shape: {matrix_a.shape}")

print(f"Size: {matrix_a.size}")

print(f"Dimension: {matrix_a.ndim}")

print("\n")
```


STEP 3: Perform operations
Extracting a second 2x2 matrix (Math and Reading scores of students 3 and 4)

1. Creating Matrix Bmatrix_b = df[['math score', 'reading score']].iloc[2:4].to_numpy()df[['math score', 'reading score']]: This selects only the two specific numerical columns from your DataFrame..iloc[2:4]: This stands for "integer location." It tells Pandas to extract specific rows based on their index numbers. In Python, index counting starts at 0, and slicing ranges are "exclusive" at the end. So, 2:4 grabs the row at index 2 (the 3rd student) and index 3 (the 4th student). It stops before index 4..to_numpy(): This strips away the Pandas formatting (row names, column headers) and converts the raw numbers into a standard Numpy array, giving you your second 2 \times 2 matrix (B).

after this using numpy operations like addition , substraction , multiplication , dot product, transpose  and inverse


```python

matrix_b = df[['math score', 'reading score']].iloc[2:4].to_numpy()

print("Step 3 : using numpy operations ")
print(f"Matrix B:\n{matrix_b}\n")

# Addition
print("Addition (A + B):\n", matrix_a + matrix_b)

# Subtraction
print("\nSubtraction (A - B):\n", matrix_a - matrix_b)

# Multiplication (Element-wise)
print("\nMultiplication (A * B):\n", matrix_a * matrix_b)

# Product (Dot Product / Matrix Multiplication)
print("\nDot Product (A @ B):\n", np.dot(matrix_a, matrix_b))

# Transpose
print("\nTranspose of A:\n", matrix_a.T)

# Inverse (Requires a square matrix)
print("\nInverse of A:\n", np.linalg.inv(matrix_a))
print("\n")

```

OPTIONAL: One-Hot Encoding (Categorical to Numerical)

In a standard Machine Learning Life Cycle, this specific operation falls squarely into the Data Preprocessing phase. Mathematical algorithms and neural networks cannot perform calculations on text strings like "male" or "female", so we must convert them into a numerical format before feeding the data into a model.


pd.get_dummies(...): This is the built-in Pandas function specifically designed to perform One-Hot Encoding. It scans a column for unique text categories and creates a brand new binary column for each unique category it finds.

df_encoded[['...']]: We look inside our newly created DataFrame and select a mix of original numerical columns (math and reading scores) alongside our brand new One-Hot Encoded columns (gender_female and gender_male).

```python
print("--- OPTIONAL: One-Hot Encoding ---")
# Converting the categorical 'gender' column ('male' / 'female') into numerical format
df_encoded = pd.get_dummies(df, columns=['gender'], dtype=int)

# Displaying the new encoded columns
print(df_encoded[['math score', 'reading score', 'gender_female', 'gender_male']].head())
```
